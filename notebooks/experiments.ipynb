```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "src").is_dir():
    ROOT = ROOT.parent
if not (ROOT / "src").is_dir():
    raise RuntimeError("Откройте ноутбук из корня проекта или notebooks")
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.training.experiments import ExperimentRegistry
```

```python
registry = ExperimentRegistry(ROOT / "runs")
print(registry.tree())
for issue in registry.issues:
    print("Внимание:", issue)
```

Пустые поля означают отсутствие данных. `notes_status=stub` отмечает незаполненное описание; введённые связи и скор при этом видны. Сравнивайте validation AIC только на одинаковых данных и протоколе оценки: совпадения одного `validation_resolution` недостаточно. Дельты автоматически не вычисляются.

```python
table = registry.table()
display(table)
```

Для сравнения конкретного запуска с его baseline укажите имя папки ниже.

```python
run_name = ""  # например, pvt_v2_b2_mixed_original
if run_name:
    selected = table.loc[table["run"] == run_name]
    if selected.empty:
        raise ValueError(f"Неизвестный запуск: {run_name}")
    baseline = selected.iloc[0]["baseline"]
    display(table.loc[table["run"].isin([run_name, baseline])])
```
